```python
X = user.columns.drop("Purchased")
y = user['Purchased']
```

```python
user_decoded = pd.get_dummies(user[X]).astype(int)
```

```python
user_decoded
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(user_decoded,y,test_size=0.2,random_state=0)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_test.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(criterion='gini',random_state=1)
model.fit(x_train,y_train)
```

```python
train_predictions = model.predict(x_train)
test_predictions = model.predict(x_test)
```

```python
from sklearn.tree import plot_tree
```

```python
plt.Figure(figsize=(50,50))
plot_tree(model,feature_names=user_decoded.columns,filled=True)
plt.show()
```
